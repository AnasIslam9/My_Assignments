# Smart Bus Ticket Booking System
Simple Python code using basic `if/else` and loops. Run the code cell and select an option from the menu.

```python
routes = [
    {"id":"R101", "from":"Karachi", "to":"Hyderabad", "time":"08:00 AM", "fare":1200, "total":40, "seats":40},
    {"id":"R102", "from":"Karachi", "to":"Sukkur", "time":"09:30 AM", "fare":2000, "total":36, "seats":36},
    {"id":"R103", "from":"Hyderabad", "to":"Mirpur Khas", "time":"11:00 AM", "fare":800, "total":32, "seats":32},
    {"id":"R104", "from":"Karachi", "to":"Lahore", "time":"07:00 PM", "fare":5000, "total":45, "seats":45},
    {"id":"R105", "from":"Sukkur", "to":"Hyderabad", "time":"02:00 PM", "fare":1500, "total":30, "seats":30}
]

bookings = []
booking_number = 1001

while True:
    print("\nSMART BUS TICKET BOOKING SYSTEM")
    print("1. View Routes")
    print("2. Search Route")
    print("3. Book Ticket")
    print("4. View Bookings")
    print("5. Cancel Booking")
    print("6. Seat Availability")
    print("7. Booking Report")
    print("8. Route Schedule")
    print("9. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        for route in routes:
            print(route["id"], route["from"], "to", route["to"],
                  "|", route["time"], "| Rs.", route["fare"],
                  "| Seats:", route["seats"], "/", route["total"])

    elif choice == "2":
        search = input("Enter route ID or city: ").lower()
        found = False

        for route in routes:
            if search == route["id"].lower() or search == route["from"].lower() or search == route["to"].lower():
                print(route["id"], route["from"], "to", route["to"],
                      "|", route["time"], "| Rs.", route["fare"],
                      "| Seats:", route["seats"])
                found = True

        if found == False:
            print("Route not found.")

    elif choice == "3":
        name = input("Passenger name: ")
        age = input("Passenger age: ")

        if name == "":
            print("Name cannot be empty.")
        elif not age.isdigit() or int(age) <= 0:
            print("Enter a valid age.")
        else:
            route_id = input("Route ID: ").upper()
            selected_route = None

            for route in routes:
                if route["id"] == route_id:
                    selected_route = route

            if selected_route == None:
                print("Route not found.")
            else:
                seats_text = input("Number of seats: ")

                if not seats_text.isdigit() or int(seats_text) <= 0:
                    print("Enter a valid number of seats.")
                elif int(seats_text) > selected_route["seats"]:
                    print("Not enough seats.")
                else:
                    seats = int(seats_text)
                    total_fare = selected_route["fare"] * seats
                    booking_id = "B" + str(booking_number)

                    booking = {
                        "id": booking_id,
                        "name": name,
                        "age": int(age),
                        "route": route_id,
                        "seats": seats,
                        "fare": total_fare,
                        "status": "Confirmed"
                    }

                    bookings.append(booking)
                    selected_route["seats"] = selected_route["seats"] - seats
                    booking_number = booking_number + 1

                    print("Booking confirmed!")
                    print("Booking ID:", booking_id)
                    print("Total fare: Rs.", total_fare)

    elif choice == "4":
        if len(bookings) == 0:
            print("No bookings found.")
        else:
            for booking in bookings:
                print(booking["id"], booking["name"], booking["route"],
                      "Seats:", booking["seats"], "Fare:", booking["fare"],
                      "Status:", booking["status"])

    elif choice == "5":
        booking_id = input("Enter booking ID: ").upper()
        found = False

        for booking in bookings:
            if booking["id"] == booking_id:
                found = True

                if booking["status"] == "Confirmed":
                    booking["status"] = "Cancelled"

                    for route in routes:
                        if route["id"] == booking["route"]:
                            route["seats"] = route["seats"] + booking["seats"]

                    print("Booking cancelled. Seats restored.")
                else:
                    print("Booking already cancelled.")

        if found == False:
            print("Booking not found.")

    elif choice == "6":
        for route in routes:
            if route["seats"] == 0:
                status = "Fully booked"
            elif route["seats"] == route["total"]:
                status = "Available"
            else:
                status = "Partially booked"

            print(route["id"], route["from"], "to", route["to"],
                  "|", route["seats"], "/", route["total"], "|", status)

    elif choice == "7":
        confirmed = 0
        cancelled = 0
        reserved_seats = 0
        revenue = 0
        full_routes = 0

        for booking in bookings:
            if booking["status"] == "Confirmed":
                confirmed = confirmed + 1
                reserved_seats = reserved_seats + booking["seats"]
                revenue = revenue + booking["fare"]
            else:
                cancelled = cancelled + 1

        for route in routes:
            if route["seats"] == 0:
                full_routes = full_routes + 1

        print("Total bookings:", len(bookings))
        print("Confirmed bookings:", confirmed)
        print("Cancelled bookings:", cancelled)
        print("Seats currently reserved:", reserved_seats)
        print("Confirmed revenue: Rs.", revenue)
        print("Fully booked routes:", full_routes)

    elif choice == "8":
        cities = ["Karachi", "Hyderabad", "Sukkur", "Mirpur Khas", "Lahore"]

        for city in cities:
            print(city)
            found = False

            for route in routes:
                if route["from"] == city:
                    print(route["id"], "to", route["to"], route["time"])
                    found = True

            if found == False:
                print("No scheduled departures.")

    elif choice == "9":
        print("Goodbye!")
        break

    else:
        print("Invalid choice.")
```
