## Question 14 — Restaurant Bill

**Problem:**  
A restaurant wants a function to calculate the final bill after adding tax and applying discount.

  - Food price
  - Quantity

  - Discount percentage

````python
def restaurant_bill(food_price, quantity, tax, discount):
    subtotal = food_price * quantity
    discount_amount = subtotal * discount / 100
    amount_after_discount = subtotal - discount_amount
    tax_amount = amount_after_discount * tax / 100
    final_amount = amount_after_discount + tax_amount
    return final_amount

bill = restaurant_bill(1000, 2, 5, 10)
print("Final Bill: Rs.", bill)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_percent, tax_percent):
    def subtotal():
        return price * quantity

    def discount():
        return subtotal() * discount_percent / 100

    def tax():
        amount = subtotal() - discount()
        return amount * tax_percent / 100

    def final_amount():
        return subtotal() - discount() + tax()

    return final_amount()

total = process_order(1500, 2, 10, 5)
print("Final Order Amount: Rs.", total)
````
